# Evaluation of the Utils class

```python
import numpy as np
from mnplib.utils import empirical_distribution
```

### Categorical Vectors

```python
x = np.array(["cat", "cat", "dog", "bird", "dog", "cat"])
summary = empirical_distribution(columns=[x], numeric=[False])
```

```python
print(summary.states)
print(summary.counts)
print(summary.probabilities)
# (1/2) * (-np.log2(1/2)) + (1/3) * (-np.log2(1/3)) + (1/6) * (-np.log2(1/6)) = 1.46
print(summary.entropy)
# 3 * (-np.log2(1/2)) + 2 * (-np.log2(1/3)) + 1 * (-np.log2(1/6)) = 8.75
print(summary.code_length)
```

### Numerical Vectors

```python
x = np.array([1.0, 1.2, 1.3, 4.8, 5.0, 5.2])
summary = empirical_distribution(columns=[x], numeric=[True], n_bins=2)
```

```python
print(summary.states)
print(summary.counts)
print(summary.probabilities)
print(summary.entropy)
print(summary.code_length)
```

### Join Empirical Distributions

```python
# (A,0), (A,0), (B,1), (B,1), (B,0), (A,0)
x = np.array(["A", "A", "B", "B", "B", "A"])
y = np.array([0, 0, 1, 1, 0, 0])
summary = empirical_distribution(columns=[x, y], numeric=[False, False])
```

```python
print(summary.states)
print(summary.counts)
print(summary.probabilities)
print(summary.entropy)
print(summary.code_length)
```

```python
# (A,0,0), (A,0,0), (B,1,0), (B,1,1), (B,0,1), (A,0,1)
x = np.array(["A", "A", "B", "B", "B", "A"])
y = np.array([0, 0, 1, 1, 0, 0])
z = np.array([0, 0, 0, 1, 1, 1])
summary = empirical_distribution(columns=[x, y, z], numeric=[False, False, False])
```

```python
print(summary.states)
print(summary.counts)
print(summary.probabilities)
print(summary.entropy)
print(summary.code_length)
```
