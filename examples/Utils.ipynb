# Empirical Distributions and Code Lengths

Use `empirical_distribution_vector(x)` for a one-dimensional vector with automatic binning. Use `empirical_distribution_array(X)` for a two-dimensional array with samples in rows and variables in columns; its default is adaptive binning. Both return an `EmpiricalSummary` with a `code_length` measured in bits.

```python
import numpy as np
from mnplib.utils import empirical_distribution_vector, empirical_distribution_array
```

### Categorical Vectors

```python
x = np.array(["cat", "cat", "dog", "bird", "dog", "cat"])
summary = empirical_distribution_vector(x, numeric=False)
```

```python
print(summary.states)
print(summary.counts)
print(summary.probabilities)
# 3 * (-np.log2(1/2)) + 2 * (-np.log2(1/3)) + 1 * (-np.log2(1/6)) = 8.75
print(summary.code_length)
```

### Numerical Vectors

```python
x = np.array([1.0, 1.2, 1.3, 4.8, 5.0, 5.2])
summary = empirical_distribution_vector(x, n_bins=2)
```

```python
print(summary.states)
print(summary.counts)
print(summary.probabilities)
print(summary.code_length)
```

### Adaptive Joint Discretization

The array function defaults to `n_bins="adaptive"`. For `d` columns, it uses `max(2, floor(2 * n_samples**(1/3) / log2(d + 1)))` bins per numeric variable. All columns count toward `d`; categorical variables retain their category codes. For one variable, adaptive and automatic binning agree.

When comparing marginal and joint code lengths in a metric, use the same explicit integer bin count for all related distributions. `Miscoding` resolves that count from the selected feature-subset size, excluding the target.

```python
x = np.linspace(0.0, 1.0, 100)
y = x ** 2
z = np.sin(np.pi * x)

single = empirical_distribution_vector(x)
print(f"K(X): {single.code_length:.4f} bits")

X = np.column_stack([x, y, z])
joint = empirical_distribution_array(X)
print(f"adaptive: {joint.n_states} joint states, {joint.code_length:.4f} bits")

joint_auto = empirical_distribution_array(X, n_bins="auto")
print(f"auto: {joint_auto.n_states} joint states, {joint_auto.code_length:.4f} bits")
```

### Joint Categorical Distributions

Set `numeric=False` for categorical data. For mixed data, pass a sequence such as `numeric=[True, False]`, with one boolean per column. A pandas DataFrame or an object array preserves mixed value types.

```python
# (A,0), (A,0), (B,1), (B,1), (B,0), (A,0)
x = np.array(["A", "A", "B", "B", "B", "A"])
y = np.array([0, 0, 1, 1, 0, 0])
X = np.array([x, y], dtype=object).T
summary = empirical_distribution_array(X, numeric=False)
```

```python
print(summary.states)
print(summary.counts)
print(summary.probabilities)
print(summary.code_length)
```

```python
# (A,0,0), (A,0,0), (B,1,0), (B,1,1), (B,0,1), (A,0,1)
x = np.array(["A", "A", "B", "B", "B", "A"])
y = np.array([0, 0, 1, 1, 0, 0])
z = np.array([0, 0, 0, 1, 1, 1])
X = np.array([x, y, z], dtype=object).T
summary = empirical_distribution_array(X, numeric=False)
```

```python
print(summary.states)
print(summary.counts)
print(summary.probabilities)
print(summary.code_length)
```
