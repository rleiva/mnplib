# A Study of Compressors Overhead

```python
import bz2
import lzma
import zlib

import numpy as np
```

Compressing the empty string

```python
len(lzma.compress(b"", preset=9))
```

```python
len(zlib.compress(b"", level=9))
```

```python
len(bz2.compress(b"", compresslevel=9))
```

Incompressibility of random strings

```python
def estimate_overhead(compressor, n_bytes, n_trials=20, random_state=42):
    rng = np.random.default_rng(random_state)
    overheads = []

    for _ in range(n_trials):
        random_bytes = rng.integers(
            0,
            256,
            size=n_bytes,
            dtype=np.uint8,
        ).tobytes()

        if compressor == "lzma":
            length =  len(lzma.compress(random_bytes, preset=9))
        elif compressor == "zlib":
            length = len(zlib.compress(random_bytes, level=9))
        else:
            length = len(bz2.compress(random_bytes, compresslevel=9))

        overheads.append(length - n_bytes)

    return int(np.median(overheads))
```

```python
estimate_overhead("lzma", 80)
```

```python
estimate_overhead("zlib", 10)
```

```python
estimate_overhead("bz2", 70)
```
